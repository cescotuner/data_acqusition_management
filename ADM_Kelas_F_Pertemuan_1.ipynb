### 3. reproducibility penting dalam akuisisi data, sertakan satu contoh kasus nyata.

Reproducibility atau keterulangan sangat penting dalam proses akuisisi data, reproducibility memungkinkan proses pengumpulan dan pengolahan data dilakukan berulang dengan langkah-langkah yang sama. Dengan adanya reproducibility, hasil yang ditampilkan dapat diperiksa dan dibandingkan sehingga menghindari terjadinya kesalahan dalam pengumpulan data. Selain itu, diperlukan dokumentasi yang baik untuk membantu orang lain memahami proser perolehan dan cara mengolah data. Penerapan reproducibility dapat dilakukan dengan menyimpan kode program, mencatat sumber data, menuliskan waktu pengambilan data, serta mendokumentasikan langkah-langkah seperti pembersihan dan perubahan data. Contoh penerepannya adalah pengumpulan data harga beras dari situs web pemerintahan untuk mengetahui perubahan harga, entah kenaikan atau penurunan. Pengambilan data dapat dilakukan menggunakan script Python yang disimpan agar bisa dijalankan kembali. Jika suatu saat hasil pengolahan data menunjukkan angka yang berbeda, proses sebelumnya dapat diperiksa untuk mengetahui penyebabnya. Namun, hasil pengambilan data tidak selalu sama karena harga beras dapat berubah atau situs web mengalami pembaruan. Oleh karena itu, data mentah, kode program, tanggal pengambilan, dan metode pengolahan perlu disimpan dengan baik. Dengan begitu, proses akuisisi data menjadi lebih teratur, transparan, dan dapat dipertanggungjawabkan.
